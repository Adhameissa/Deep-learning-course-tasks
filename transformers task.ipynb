```
import numpy as np

tokens = ["What", "are", "the", "symptoms", "of", "diabetes"]

np.random.seed(42)
embedding_dim = 8
embeddings = np.random.rand(len(tokens), embedding_dim)

W_Q = np.random.rand(embedding_dim, embedding_dim)
W_K = np.random.rand(embedding_dim, embedding_dim)
W_V = np.random.rand(embedding_dim, embedding_dim)

Q = embeddings @ W_Q
K = embeddings @ W_K
V = embeddings @ W_V

dk = K.shape[1]
scores = (Q @ K.T) / np.sqrt(dk)

def softmax(x):
    exp_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return exp_x / np.sum(exp_x, axis=-1, keepdims=True)

attention_weights = softmax(scores)

output = attention_weights @ V

print("Tokens:", tokens)
print("\nAttention Scores:\n", scores)
print("\nAttention Weights (Softmax):\n", attention_weights)
print("\nSelf-Attention Output:\n", output)
```
